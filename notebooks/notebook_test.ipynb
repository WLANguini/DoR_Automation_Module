```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
tick = "AAPL"
ticker = yf.Ticker(tick)
data = ticker.history(period = "max")
print(data)
```

```python
df = pd.DataFrame(data)
df = df.reset_index()
df
```
