```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
```

Daily DoR for AAPL

```python
tick = "AAPL"
ticker = yf.Ticker(tick)
data = ticker.history(period = "max")
print(data)
```

```python
df = pd.DataFrame(data)
df = df.reset_index()
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.to_csv("AAPL_DoR_Daily")
```

```python
df = df.drop(columns= "Dividends")
df = df.drop(columns= "Stock Splits")
df = df.drop(columns= "Volume")
```

```python
df = df.sort_values(by= "Date", ascending=False)
df
```

```python
df["O-C Returns"] = ((df["Close"] - df["Open"])/df["Open"])*100
df["H-L Returns"] = ((df["High"] - df["Low"])/df["Low"])*100
df
```

```python
df['C-C Returns'] = df['Close'].pct_change() * 100
df
```
